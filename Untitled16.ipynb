```python
le_glucose = LabelEncoder()

input_n['glucose_level_n'] = le_glucose.fit_transform(input_n['glucose_level'])

input_n = input_n.drop('glucose_level', axis=1)

input_n.dtypes
```

```python
X = input_n.drop('has_diabetes', axis=1)
y = input_n['has_diabetes']
```
